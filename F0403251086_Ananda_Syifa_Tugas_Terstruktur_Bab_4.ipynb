```python
%cd sistem-panen

# Membuat berkas Python main.py berisi fungsi penghitung total panen
code = """def hitung_total_panen(jumlah_kg, harga_per_kg):
    return jumlah_kg * harga_per_kg
"""
with open("main.py", "w") as f:
    f.write(code)

# Cek status, add, dan commit pertama (Soal a)
!git status
!git add main.py
!git commit -m "Inisialisasi commit: menambahkan program penghitung total hasil panen"
```

```python
# Membuat dan pindah ke branch baru bernama fitur-diskon
!git checkout -b fitur-diskon

# Menambahkan fungsi diskon ke dalam berkas main.py
discount_code = """

def hitung_diskon(total_harga, diskon_persen):
    return total_harga - (total_harga * diskon_persen / 100)
"""
with open("main.py", "a") as f:
    f.write(discount_code)

# Cek isi file main.py yang sudah diperbarui
with open("main.py", "r") as f:
    print(f.read())

# Commit perubahan pada branch fitur-diskon (Soal b)
!git add main.py
!git commit -m "Menambahkan fungsi hitung_diskon pada branch fitur-diskon"
```
